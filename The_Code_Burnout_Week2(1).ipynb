```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
missing = df.isnull().sum()

print("Missing Values in Each Column:")
display(missing[missing > 0].sort_values(ascending=False))
```

```python
print("FIRST IMPRESSION REPORT")
print("-----------------------")

print("1. Dataset Name: OSMI Mental Health in Tech Survey 2016")

print("2. Total Responses:", df.shape[0])

print("3. Total Features:", df.shape[1])

print("4. Duplicate Rows:", df.duplicated().sum())

print("5. Numerical Columns:",
      len(df.select_dtypes(include=np.number).columns))

print("6. Categorical Columns:",
      len(df.select_dtypes(include='object').columns))

print("7. Total Missing Values:",
      df.isnull().sum().sum())

print("8. Dataset Memory Usage:",
      round(df.memory_usage(deep=True).sum()/1024, 2), "KB")

print("\nWeek 2 Data Understanding Completed!")
```

```python
with open("Week2_First_Impression_Report.txt", "w") as f:
    f.write("FIRST IMPRESSION REPORT\n")
    f.write("OSMI Mental Health in Tech Survey 2016\n\n")
    f.write(f"Total Rows: {df.shape[0]}\n")
    f.write(f"Total Columns: {df.shape[1]}\n")
    f.write(f"Duplicate Rows: {df.duplicated().sum()}\n")
    f.write(f"Missing Values: {df.isnull().sum().sum()}\n")
    f.write(f"Numerical Columns: {len(df.select_dtypes(include=np.number).columns)}\n")
    f.write(f"Categorical Columns: {len(df.select_dtypes(include='object').columns)}\n")

print("Report saved successfully!")
```

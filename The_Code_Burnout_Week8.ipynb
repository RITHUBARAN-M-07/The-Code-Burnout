WEEK 8 – Interactive Dashboard using Plotly

```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
age_col = 'What is your age?'

gender_col = 'What is your gender?'

treatment_col = 'Have you ever sought treatment for a mental health issue from a mental health professional?'

family_col = 'Do you have a family history of mental illness?'

treated_col = 'If you have a mental health issue, do you feel that it interferes with your work when being treated effectively?'

not_treated_col = 'If you have a mental health issue, do you feel that it interferes with your work when NOT being treated effectively?'

required_columns = [
    age_col,
    gender_col,
    treatment_col,
    family_col,
    treated_col,
    not_treated_col
]

print("Required Columns Available:")
display(df[required_columns].head())
```

```python
dashboard_data = df[required_columns].copy()

dashboard_data[age_col] = pd.to_numeric(
    dashboard_data[age_col],
    errors='coerce'
)

dashboard_data.loc[
    (dashboard_data[age_col] < 18) |
    (dashboard_data[age_col] > 100),
    age_col
] = np.nan

dashboard_data = dashboard_data.dropna(subset=[age_col])

for col in required_columns[1:]:
    dashboard_data[col] = (
        dashboard_data[col]
        .fillna('Unknown')
        .astype(str)
        .str.strip()
    )

print("Data Cleaning Completed!")
print("Cleaned Dataset Shape:", dashboard_data.shape)

display(dashboard_data.head())
```

```python
fig1 = px.histogram(
    dashboard_data,
    x=age_col,
    nbins=20,
    title="1. Age Distribution of Survey Participants",
    labels={age_col: "Age"}
)

fig1.update_layout(
    xaxis_title="Age",
    yaxis_title="Number of Participants",
    template="plotly_white"
)

fig1.show()
```

```python
gender_counts = (
    dashboard_data[gender_col]
    .value_counts()
    .reset_index()
)

gender_counts.columns = ["Gender", "Count"]

fig2 = px.bar(
    gender_counts,
    x="Gender",
    y="Count",
    title="2. Gender Distribution",
    text="Count",
    color="Gender"
)

fig2.update_layout(
    xaxis_title="Gender",
    yaxis_title="Number of Participants",
    template="plotly_white"
)

fig2.show()
```

```python
treatment_counts = (
    dashboard_data[treatment_col]
    .value_counts()
    .reset_index()
)

treatment_counts.columns = ["Treatment", "Count"]

fig3 = px.pie(
    treatment_counts,
    names="Treatment",
    values="Count",
    title="3. Mental Health Treatment Distribution",
    hole=0.3
)

fig3.show()
```

```python
family_counts = (
    dashboard_data[family_col]
    .value_counts()
    .reset_index()
)

family_counts.columns = ["Family History", "Count"]

fig4 = px.bar(
    family_counts,
    x="Family History",
    y="Count",
    title="4. Family History of Mental Illness",
    text="Count",
    color="Family History"
)

fig4.update_layout(
    xaxis_title="Family History",
    yaxis_title="Number of Participants",
    template="plotly_white"
)

fig4.show()
```

```python
treated_counts = (
    dashboard_data[treated_col]
    .value_counts()
    .rename_axis("Response")
    .reset_index(name="Count")
)

treated_counts["Treatment Status"] = "When Treated Effectively"

not_treated_counts = (
    dashboard_data[not_treated_col]
    .value_counts()
    .rename_axis("Response")
    .reset_index(name="Count")
)

not_treated_counts["Treatment Status"] = "When Not Treated Effectively"

work_data = pd.concat(
    [treated_counts, not_treated_counts],
    ignore_index=True
)

fig5 = px.bar(
    work_data,
    x="Response",
    y="Count",
    color="Treatment Status",
    barmode="group",
    title="5. Work Interference and Treatment Status"
)

fig5.update_layout(
    xaxis_title="Work Interference Response",
    yaxis_title="Number of Participants",
    template="plotly_white"
)

fig5.show()
```

```python
import plotly.io as pio

figures = [fig1, fig2, fig3, fig4, fig5]

html_parts = []

for i, fig in enumerate(figures):
    html_parts.append(
        pio.to_html(
            fig,
            full_html=False,
            include_plotlyjs="cdn" if i == 0 else False
        )
    )

dashboard_html = """
<html>
<head>
<title>The Code Burnout Dashboard</title>
<style>
body {
    font-family: Arial, sans-serif;
    background-color: #f4f4f4;
    margin: 25px;
}
h1 {
    text-align: center;
    color: #222;
}
h2 {
    text-align: center;
}
</style>
</head>
<body>

<h1>THE CODE BURNOUT</h1>

<h2>AI-Based Analysis of Mental Health in the Tech Industry</h2>

""" + "\n".join(html_parts) + """

</body>
</html>
"""

with open("The_Code_Burnout_Dashboard.html", "w") as f:
    f.write(dashboard_html)

print("Interactive Dashboard Created Successfully!")
print("File: The_Code_Burnout_Dashboard.html")
```

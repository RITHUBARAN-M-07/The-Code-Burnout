```python
import pandas as pd
import numpy as np
from google.colab import files

# Step 1: Upload dataset
uploaded = files.upload()

# Step 2: Load dataset
file_name = next(iter(uploaded))
clean = pd.read_csv(file_name)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", clean.shape)

# Step 3: Select features
selected_columns = [
    'What is your age?',
    'What is your gender?',
    'Do you have a family history of mental illness?',
    'If you have a mental health issue, do you feel that it interferes with your work when being treated effectively?',
    'If you have a mental health issue, do you feel that it interferes with your work when NOT being treated effectively?',
    'Have you ever sought treatment for a mental health issue from a mental health professional?'
]

# Step 4: Check columns
missing_columns = [
    col for col in selected_columns
    if col not in clean.columns
]

if len(missing_columns) == 0:
    feature_data = clean[selected_columns].copy()
    print("Feature Selection Successful!")
    print("Feature Shape:", feature_data.shape)
    display(feature_data.head())
else:
    print("Missing Columns:", missing_columns)
```

```python
age_col = 'What is your age?'

feature_data[age_col] = pd.to_numeric(
    feature_data[age_col], errors='coerce'
)

feature_data[age_col] = feature_data[age_col].fillna(
    feature_data[age_col].median()
)

categorical_cols = feature_data.select_dtypes(
    include='object'
).columns

for col in categorical_cols:
    feature_data[col] = feature_data[col].fillna('Unknown')

print("Missing Values After Handling:")
print(feature_data.isnull().sum())
```

```python
encoded_data = pd.get_dummies(
    feature_data,
    columns=[
        'What is your gender?',
        'Do you have a family history of mental illness?',
        'If you have a mental health issue, do you feel that it interferes with your work when being treated effectively?',
        'If you have a mental health issue, do you feel that it interferes with your work when NOT being treated effectively?',
        'Have you ever sought treatment for a mental health issue from a mental health professional?'
    ],
    dtype=int
)

print("Categorical Encoding Completed!")
print("Encoded Dataset Shape:", encoded_data.shape)

display(encoded_data.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

age_col = 'What is your age?'

encoded_data[age_col] = scaler.fit_transform(
    encoded_data[[age_col]]
).ravel()

print("Feature Scaling Completed!")
display(encoded_data.head())
```

```python
encoded_data.to_csv(
    "Week6_Feature_Engineered_Data.csv",
    index=False
)

print("Week 6 Dataset Saved Successfully!")
```
